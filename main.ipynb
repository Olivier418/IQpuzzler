```python
import os

from classes import SolutionBook
from plotting.plot_solve_timeline import plot_solve_timeline
from serialization import save_solutions, load_solutions, load_game
from plotting.plot_difficultyspace import plot_difficulty_space
import matplotlib.pyplot as plt
```

Let's take a look at puzzle x in the main puzzles booklet.

```python
IQpuzzler = load_game("games/IQpuzzler")
main_puzzle_book = IQpuzzler['main_puzzles']

puzzle = main_puzzle_book['65']
puzzle.print_to_terminal()
puzzle.solve(disp = True)
```

We see that there are actually multiple solutions! Let's investigate the other puzzles in the booklet as well. We can compute their solutions in one go by using the SolutionBook.from_puzzlebook method. We then save these solutions to a file for later investigation.

```python
solution_book = SolutionBook.from_puzzlebook(main_puzzle_book, disp=True)
SOLUTION_DIR = os.path.join("solutions")
solution_folder = save_solutions(solution_book, SOLUTION_DIR)
```

Now, we can load the solutions and analyze them! Let's first look at the number of solutions vs empty spaces per puzzle!

```python
solution_book = load_solutions(solution_folder)

plot_difficulty_space(solution_book)
plt.show()
```

Alternatively, we can investigate how long it took to solve each puzzle, and how many solutions were found.

```python
plot_solve_timeline(solution_book)
plt.show()
```

Apparently, some supposedly difficult puzzles have a huge amount of solutions!

The developers at smartgames know this. They have released an updated version, called the IQpuzzlerPRO, which supposedly has unique solutions only! Let's investigate. The solutions are already computed, we just load them and plot the results.

```python
solution_folder = os.path.join(SOLUTION_DIR,"IQpuzzlerPRO_main_puzzles_solutions")
solution_book = load_solutions(solution_folder)
plot_solve_timeline(solution_book)
plt.show()
```
